```python
import tkinter
import pymysql
from tkinter import *
from tkinter import ttk
from tkinter import messagebox
t=tkinter.Tk()
t.geometry("700x600")
def finddata():
    db=pymysql.connect(host='localhost',user='root',password='root',database='my56')
    cur=db.cursor()
    xa=int(ea.get())
    sql="select name,city,salary,grade from staff where staffid=%s"
    cur.execute(sql,(xa))
    data=cur.fetchone()
    eb.insert(0,data[0])
    ec.insert(0,data[0])
    ed.insert(0,data[0])
    ee.insert(0,data[0])
    db.close()
a=Label(t,text='Staffid')
a.place(x=50,y=50)
ea=Entry(t,width=20)
ea.place(x=400,y=50)
b=Label(t,text='Name')
b.place(x=50,y=90)
eb=Entry(t,width=20)
eb.place(x=400,y=90)
bt=Button(t,text='Find',command=finddata)
bt.place(x=550,y=50)
eb=Entry(t,width=20)
eb.place(x=400,y=90)
c=Label(t,text='City')
c.place(x=50,y=130)
ec=Entry(t,width=20)
ec.place(x=400,y=130)
d=Label(t,text='Salary')
d.place(x=50,y=170)
ed=Entry(t,width=20)
ed.place(x=400,y=170)
d=Label(t,text='Grade')
d.place(x=50,y=200)
ee=Entry(t,width=20)
ee.place(x=400,y=200)
t.mainloop()
```
